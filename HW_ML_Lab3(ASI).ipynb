```python
X = df.drop(columns=['y'])
y = df['y'].map({'yes': 1, 'no': 0})
```

```python
X.head()
```

```python
y.head()
```

```python
binary_cols = ['default', 'housing', 'loan']
for col in binary_cols:
    X[col] = X[col].map({'yes': 1, 'no': 0})
```

```python
X['education'] = X['education'].map({
    'unknown': 0,
    'primary': 1,
    'secondary': 2,
    'tertiary': 3
})
```

```python
X.head()
```

```python
from sklearn.model_selection import train_test_split
x_train, x_test, y_train, y_test = train_test_split(X, y, test_size = 0.2, random_state = 2)
```

```python
from sklearn.linear_model import LogisticRegression
```

```python
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
import pandas as pd

# Identify categorical columns that are still objects
categorical_cols_to_encode = ['job', 'marital', 'contact', 'month', 'poutcome']

# Apply one-hot encoding to x_train and x_test
x_train_encoded = pd.get_dummies(x_train, columns=categorical_cols_to_encode, drop_first=True)
x_test_encoded = pd.get_dummies(x_test, columns=categorical_cols_to_encode, drop_first=True)

# Align columns: This ensures both train and test sets have the same columns
# which is crucial if certain categories are present in one set but not the other.
x_test_encoded = x_test_encoded.reindex(columns=x_train_encoded.columns, fill_value=0)

scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(x_train_encoded)
X_test_scaled = scaler.transform(x_test_encoded)

lg_model = LogisticRegression(class_weight='balanced', max_iter=1000)
lg_model.fit(X_train_scaled, y_train)
```

```python
y_pred = lg_model.predict(X_test_scaled)
```

```python
from sklearn.metrics import accuracy_score , precision_score, recall_score, f1_score , confusion_matrix
```

```python
accuracy = accuracy_score(y_test, y_pred)
precision = precision_score(y_test, y_pred)
recall = recall_score(y_test, y_pred)
f1 = f1_score(y_test, y_pred)
conf_matrix = confusion_matrix(y_test, y_pred)
```

```python
print("Accuracy: ", accuracy)
print("Precision: ", precision)
print("Recall: ", recall)
print("F1 Score: ", f1)
print("Confusion Matrix: \n", conf_matrix)
```
