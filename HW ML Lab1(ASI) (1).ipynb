```python
from sklearn.preprocessing import Binarizer

binarizer = Binarizer(threshold=df['medv'].mean())

df_binarized_median_house_value = pd.DataFrame(
    binarizer.fit_transform(df[['medv']]),
    columns=['median_house_value']
)
```

```python
df['median_house_value_binarized'] = df_binarized_median_house_value['median_house_value']
print(df['median_house_value_binarized'].unique())
```
